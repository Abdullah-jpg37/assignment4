### Q5 `numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# File se numbers read karke unka sum nikalne ka code
total_sum = 0

with open("numbers.txt", "r") as file:
    for line in file:
        # strip() line ke aakhir se newline character (\n) ko hatata hai
        number = int(line.strip())
        total_sum += number

print(f"Tamam numbers ka kul sum (Total) hai: {total_sum}")
```

### Q6 Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
file_name = "numbers.txt"

line_count = 0

try:
    with open(file_name, "r") as file:
        for line in file:
            line_count += 1
            
    print(f"File '{file_name}' mein total lines hain: {line_count}")

except FileNotFoundError:
    print(f"Error: '{file_name}' naam ki koi file nahi mili.")
```

### Q7 (Medium)`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
# 1. Reading the file safely using 'with open'
with open("data.txt", "r") as file:
    content = file.read()
    print("--- File Content ---")
    print(content)
    print("-------------------\n")

# 2. Correct way to print the explanation so it looks clean
explanation = """FAIDA (ADVANTAGE) OF 'with' STATEMENT:

1. AUTOMATIC CLOSE: It automatically closes the file as soon as the 
   indented block ends. You don't need to write 'file.close()' manually.
   
2. RESOURCE MANAGEMENT: It releases system memory and prevents leaks, 
   ensuring other programs can access the file without blockage.
   
3. EXCEPTION SAFETY: Even if an error/crash occurs inside the 'with' 
   block, Python guarantees that the file will be safely closed."""

print(explanation)
```

### Q14 Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
def check_division(a, b):
    try:
        print(f"\n--- Calculation Shuru: {a} ÷ {b} ---")
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        # Yeh block tab chalega jab zero se divide karne par error aayega
        print("Error: Aap kisi bhi number ko zero (0) se divide nahi kar sakte!")
    finally:
        # Yeh block HUMESHA chalega
        print("Finally Block: Yeh line har haal mein execute hogi.")

# Case 1: Bina kisi error ke (No Error)
check_division(10, 2)

# Case 2: Error ke sath (Division by Zero Error)
check_division(10, 0)
```

### Q15 Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

```python
try:
    # 1. List banayi
    my_list = [10, 20, 30]
    
    print("List access karne ki koshish kar rahe hain...")
    # 2. Invalid index (5) ko access karne ki koshish (Yahan error aayega)
    print(my_list[5]) 
    
except IndexError:
    # 3. IndexError ko catch kiya gaya
    print("Error Catch Hua: Aap jo index access kar rahe hain wo list ki range se bahar (out of range) hai!")

finally:
    # 4. Finally block jo humesha chalega
    print("Finally Block: Execution khatam.")
```

### Q17 Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
def safe_divide():
    try:
        num1 = input("Enter First Number:")
        num2 = input("Enter Second Number:")
        
        n1 = int(num1)
        n2 = int(num2)
        
        result = n1 / n2
        print(f"Result: {result}")
        
    except ValueError:
        print("ValueError Catch Hua: Kripya sirf valid numbers (digits) hi enter karein!")
        
    except ZeroDivisionError:
        print("ZeroDivisionError Catch Hua: Aap kisi bhi number ko 0 se divide nahi kar sakte!")

# --- Program ko check karne ke alag alag tareeqe ---

print("--- Case 1: Agar galat input (text) dalein ---")
safe_divide()

print("\n--- Case 2: Agar 0 se divide karein ---")
safe_divide()
```

### Q18 `else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.


```python
def check_even_number(num):
    try:
        print(f"\n--- Checking Number: {num} ---")
        number = int(num)
        
    except ValueError:
        print("Error: Kripya sirf ek valid number dalein!")
        
    else:
        print("Success: Koi error nahi aaya!")
        if number % 2 == 0:
            print(f"{number} ek Even (Juft) number hai.")
        else:
            print(f"{number} ek Odd (Taq) number hai.")

check_even_number("10")
check_even_number("abc")
```

### Q19 `raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def validate_age(age):
    print(f"\nAge check ki ja rahi hai: {age}")
    if age < 18:
        raise ValueError("Age must be 18 or older")
    
    print("Validation Kamyab: Aap register kar sakte hain!")

try:
    validate_age(22)
except ValueError as error:
    print(f"Error pakda gaya: {error}")

try:
    validate_age(15)
except ValueError as error:
    print(f"Error pakda gaya: {error}")
```

### Q20 File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
file_ptr = None 
try:
    print("File open karne ki koshish ki ja rahi hai...")
    file_ptr = open("non_existent_file.txt", "r")
    
    content = file_ptr.read()
    print("File ka data:", content)

except FileNotFoundError:
    print("Error Catch Hua: Maazrat, 'non_existent_file.txt' naam ki koi file nahi mili!")

finally:
    if file_ptr is not None:
        file_ptr.close()
        print("Finally Block: File ko safely close kar diya gaya hai.")
    else:
        print("Finally Block: File khuli hi nahi thi, isliye close karne ki zaroorat nahi pari.")
```
